# Agglomerative (Hierarchical) Clustering


This notebook performs Agglomerative Clustering using Ward linkage, evaluates cluster quality, and produces a dendrogram and PCA 2D visualization.

## 1. Import libraries

```python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

warnings.filterwarnings("ignore")
RANDOM_STATE = 42
```

## 2. Load dataset

```python
DATA_PATH = "../dataset/clustering_patients.csv"
if not os.path.exists(DATA_PATH):
    if os.path.exists("dataset/clustering_patients.csv"):
        DATA_PATH = "dataset/clustering_patients.csv"
    elif os.path.exists("/mnt/data/clustering_patients(3).csv"):
        DATA_PATH = "/mnt/data/clustering_patients(3).csv"
    elif os.path.exists("/mnt/data/clustering_patients.csv"):
        DATA_PATH = "/mnt/data/clustering_patients.csv"

df = pd.read_csv(DATA_PATH)
print("Dataset shape:", df.shape)
display(df.head())
```

## 3. Dataset overview

```python
df.info()
print("\\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))
print("\\nDuplicate rows:", df.duplicated().sum())
display(df.describe().T)
```

## 4. Handle missing values and scale features

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
if len(numeric_columns) != df.shape[1]:
    print("Warning: non-numeric columns are excluded from clustering.")

X_raw = df[numeric_columns].copy()
imputer = SimpleImputer(strategy="median")
X_imputed = pd.DataFrame(imputer.fit_transform(X_raw), columns=numeric_columns, index=df.index)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

print("Missing values after imputation:", int(X_imputed.isnull().sum().sum()))
print("Scaled matrix shape:", X_scaled.shape)
```

## 5. Exploratory data analysis

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()
for i, col in enumerate(numeric_columns):
    axes[i].hist(X_imputed[col], bins=25, edgecolor="black")
    axes[i].set_title(f"Distribution of {col}")
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("Frequency")
for j in range(len(numeric_columns), len(axes)):
    axes[j].axis("off")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))
corr = X_imputed.corr()
plt.imshow(corr, aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(range(len(numeric_columns)), numeric_columns, rotation=45, ha="right")
plt.yticks(range(len(numeric_columns)), numeric_columns)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

## 6. Fit Agglomerative Clustering

```python
# Use the same initial cluster count as the K-Means notebook for easier comparison.
# Adjust this after reviewing the dendrogram if appropriate.
N_CLUSTERS = 3
agg_model = AgglomerativeClustering(n_clusters=N_CLUSTERS, linkage="ward")
labels = agg_model.fit_predict(X_scaled)

clustered_df = X_imputed.copy()
clustered_df["Agglomerative_Cluster"] = labels
display(pd.Series(labels).value_counts().sort_index().to_frame("Count"))
```

## 7. Evaluate Agglomerative Clustering

```python
metrics = pd.DataFrame([{
    "Algorithm": "Agglomerative (Ward)",
    "Number of Clusters": len(np.unique(labels)),
    "Silhouette Score": silhouette_score(X_scaled, labels),
    "Davies-Bouldin Index": davies_bouldin_score(X_scaled, labels),
    "Calinski-Harabasz Score": calinski_harabasz_score(X_scaled, labels)
}])
display(metrics.round(4))
```

**Metric interpretation:** Silhouette and Calinski–Harabasz are generally better when higher; Davies–Bouldin is better when lower. Interpret them together.

## 8. PCA 2D visualization

```python
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, s=22, alpha=0.75)
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% variance)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% variance)")
plt.title("PCA 2D Plot — Agglomerative Clusters")
plt.colorbar(scatter, label="Cluster")
plt.tight_layout()
plt.show()
```

## 9. Hierarchical dendrogram

```python
# Use a reproducible sample to keep the dendrogram readable and efficient.
from scipy.cluster.hierarchy import linkage, dendrogram

DENDROGRAM_SAMPLE_SIZE = min(250, len(X_scaled))
rng = np.random.default_rng(RANDOM_STATE)
sample_indices = np.sort(rng.choice(len(X_scaled), size=DENDROGRAM_SAMPLE_SIZE, replace=False))
X_dendro = X_scaled[sample_indices]

Z = linkage(X_dendro, method="ward")
plt.figure(figsize=(15, 6))
dendrogram(Z, no_labels=True, color_threshold=None)
plt.title(f"Hierarchical Clustering Dendrogram (Ward; sample n={DENDROGRAM_SAMPLE_SIZE})")
plt.xlabel("Sample observations")
plt.ylabel("Ward distance")
plt.tight_layout()
plt.show()
```

**Dendrogram note:** The dendrogram uses a reproducible sample for readability; the labels and metrics above use the full dataset.
