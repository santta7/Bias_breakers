# Bagging Classifier - Diabetes Classification

This notebook follows the same workflow as the earlier classification notebooks and uses the same diabetes dataset. It includes EDA, preprocessing, training, evaluation, visualisations, and result saving.


```python
# Importing libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve
)
```

```python
# Load the dataset

df = pd.read_csv("../dataset/diabetes_prediction_dataset_2000.csv")
df.head()
```

```python
# Basic dataset checking

print("Dataset shape:", df.shape)
print("\nDataset information:")
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Removing duplicates

before = len(df)
df = df.drop_duplicates()
after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

```python
# Checking and treating missing values safely
from pandas.api.types import is_numeric_dtype

print("Missing values before treatment:")
print(df.isnull().sum())

for col in df.columns:
    if is_numeric_dtype(df[col]):
        median_value = df[col].median()
        if pd.notna(median_value):
            df[col] = df[col].fillna(median_value)
    else:
        mode_values = df[col].mode(dropna=True)
        if not mode_values.empty:
            df[col] = df[col].fillna(mode_values.iloc[0])
        else:
            df[col] = df[col].fillna("Unknown")

print("\nMissing values after treatment:")
print(df.isnull().sum())
print("Total remaining missing values:", int(df.isnull().sum().sum()))
```

```python
# Check and treat outliers using the IQR method

outlier_cols = ["bmi", "HbA1c_level", "blood_glucose_level"]
outlier_summary = []
for col in outlier_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Feature": col,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Outliers Found": count
    })

    df[col] = df[col].clip(lower=lower, upper=upper)

pd.DataFrame(outlier_summary)
```

## Feature Engineering

The engineered feature combines blood glucose level and HbA1c level because both features describe blood sugar related information. Their interaction may provide additional information for the diabetes classification model.


```python
# Create an engineered feature

df["glucose_hba1c_interaction"] = (df["blood_glucose_level"] * df["HbA1c_level"])
```

```python
# Separate features and target

x = df.drop("diabetes", axis=1)
y = df["diabetes"]

print("Features:")
print(x.head())
print("\nTarget:")
print(y.head())
```

```python
# Identify numerical and categorical features
from pandas.api.types import is_numeric_dtype
num_features = [col for col in x.columns if is_numeric_dtype(x[col])]
cat_features = [col for col in x.columns if col not in num_features]
print("Numerical features:")
print(num_features)
print("\nCategorical features:")
print(cat_features)
```

```python
# Preprocessing
# Standardize numeric features and one-hot encode categorical features.
# The preprocessing object is part of the Pipeline so it is fitted on training folds only during CV.
try:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:  # compatibility with older scikit-learn versions
    encoder = OneHotEncoder(handle_unknown="ignore", sparse=False)

prep = ColumnTransformer(
    transformers=[
        ("numerical", StandardScaler(), num_features),
        ("categorical", encoder, cat_features)
    ],
    remainder="drop"
)
```

```python
# Stratified train-test split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training set:", x_train.shape)
print("Testing set:", x_test.shape)
```

## Model Training and Hyperparameter Tuning

Bagging Classifier is trained in a pipeline so encoding and scaling are fitted within each training fold. GridSearchCV uses stratified 5-fold cross-validation on the training split only.

```python
# Create Bagging Classifier
bag_model = Pipeline(
    steps=[
        ("preprocessing", prep),
        ("classifier", BaggingClassifier(estimator=DecisionTreeClassifier(random_state=42), n_estimators=50, random_state=42, n_jobs=-1))
    ]
)

bag_model
```

```python
# Tune hyperparameters using 5-fold stratified cross-validation
# Tune the number of decision-tree estimators
from sklearn.model_selection import GridSearchCV, StratifiedKFold
param_grid = {"classifier__n_estimators": [25, 50, 100], "classifier__estimator__max_depth": [None, 5, 10]}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
grid_search = GridSearchCV(bag_model, param_grid, scoring="f1_weighted", cv=cv, n_jobs=-1, verbose=1)
grid_search.fit(x_train, y_train)
bag_tuned_model = grid_search.best_estimator_
print("Best parameters:", grid_search.best_params_)
print("Best mean CV weighted F1:", round(grid_search.best_score_, 4))
bag_model = bag_tuned_model
bag_model
```

```python
# Final model fit and test-set predictions
bag_model.fit(x_train, y_train)
bag_pred = bag_model.predict(x_test)
bag_prob = bag_model.predict_proba(x_test)[:, 1] if hasattr(bag_model, "predict_proba") else bag_model.decision_function(x_test)
print("Bagging Classifier trained and predictions generated.")
print("First 10 predictions:", bag_pred[:10])
```

```python
# Calculate evaluation metrics on the held-out test set
bag_accuracy = accuracy_score(y_test, bag_pred)
bag_precision = precision_score(y_test, bag_pred, average="weighted", zero_division=0)
bag_recall = recall_score(y_test, bag_pred, average="weighted", zero_division=0)
bag_f1 = f1_score(y_test, bag_pred, average="weighted", zero_division=0)
bag_auc = roc_auc_score(y_test, bag_prob)
print("Bagging Classifier Performance (test set)")
print("Accuracy:", round(bag_accuracy, 4))
print("Weighted Precision:", round(bag_precision, 4))
print("Weighted Recall:", round(bag_recall, 4))
print("Weighted F1:", round(bag_f1, 4))
print("ROC-AUC:", round(bag_auc, 4))
```

```python
# Classification report
print(classification_report(y_test, bag_pred, target_names=["No Diabetes", "Diabetes"], zero_division=0))
```

```python
# Confusion matrix
cm_bag = confusion_matrix(y_test, bag_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm_bag, annot=True, fmt="d", cmap="Blues", xticklabels=["No Diabetes", "Diabetes"], yticklabels=["No Diabetes", "Diabetes"])
plt.title("Bagging Classifier - Confusion Matrix")
plt.xlabel("Predicted label")
plt.ylabel("Actual label")
plt.tight_layout()
plt.show()
```

```python
# ROC curve
bag_fpr, bag_tpr, _ = roc_curve(y_test, bag_prob)
plt.figure(figsize=(8, 6))
plt.plot(bag_fpr, bag_tpr, label=f"Bagging Classifier (AUC = {bag_auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Random classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Bagging Classifier - ROC Curve")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Save results without overwriting results from other algorithms
import os
os.makedirs("../results", exist_ok=True)
results_path = "../results/classification_results.csv"
new_row = pd.DataFrame([{
    "Model": "Bagging Classifier",
    "Accuracy": bag_accuracy,
    "Precision": bag_precision,
    "Recall": bag_recall,
    "Weighted F1": bag_f1,
    "ROC-AUC": bag_auc
}])
if os.path.exists(results_path):
    saved = pd.read_csv(results_path)
    saved = saved[saved["Model"] != "Bagging Classifier"]
    saved = pd.concat([saved, new_row], ignore_index=True)
else:
    saved = new_row
saved.to_csv(results_path, index=False)
print("Saved results to:", results_path)
saved.sort_values("Weighted F1", ascending=False).reset_index(drop=True)
```
